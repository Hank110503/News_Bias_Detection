```python
import os
import re
import requests
import pandas as pd
from urllib.parse import urlparse
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

CSV_PATH = r"E:/GDELP/task_3/mideast&asia.csv"
OUT_DIR = r"E:/GDELP/task_3/mideast&asia_imgs"  # 你可以改到任意目录
COLUMN_NAME = "image_url"

proxies = {
    "http": "socks5://127.0.0.1:7890",
    "https": "socks5://127.0.0.1:7890"
}

# 创建输出文件夹
os.makedirs(OUT_DIR, exist_ok=True)

# 设置 requests session 和重试策略
session = requests.Session()
# session.proxies = proxies
# session.headers.update({
#     "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/124.0.0.0 Safari/537.36"
# })

retries = Retry(total=3, backoff_factor=0.5, status_forcelist=(500, 502, 503, 504))
adapter = HTTPAdapter(max_retries=retries)
session.mount("http://", adapter)
session.mount("https://", adapter)

# 读取 CSV（把所有字段读作 str 以避免类型问题）
df = pd.read_csv(CSV_PATH, dtype=str, encoding="utf-8", low_memory=False)
if COLUMN_NAME not in df.columns:
    raise ValueError(f"CSV 中找不到列: {COLUMN_NAME}")

# 匹配常见图片扩展（支持带参数的情况）
ext_pattern = re.compile(r'\.(jpg|jpeg|png|gif)(?=[\?\#\/]|$)', flags=re.IGNORECASE)

def sanitize_filename(name: str) -> str:
    # 允许字母数字、下划线、连字符；其它字符替换为下划线
    name = re.sub(r'[^A-Za-z0-9_\-]', '_', name)
    # 防止名字过长
    return name[:200]

def unique_path(folder: str, base: str, ext: str) -> str:
    candidate = f"{base}{ext}"
    i = 1
    while os.path.exists(os.path.join(folder, candidate)):
        candidate = f"{base}_{i}{ext}"
        i += 1
    return os.path.join(folder, candidate)

def get_name_and_ext_from_url(url: str):
    """
    试图从 URL 中提取扩展名与 base name（扩展名前的最后一段路径）。
    如果未匹配到扩展名，返回 (None, None) 让调用方去用 Content-Type 或默认扩展处理。
    """
    m = ext_pattern.search(url)
    if m:
        ext = "." + m.group(1).lower()
        # base_part 是从 URL 开头到扩展名前的部分，方便取最后一段作为名字
        base_part = url[:m.start()]
        parsed = urlparse(base_part)
        base_name = os.path.basename(parsed.path) or None
        return base_name, ext
    else:
        # 未找到扩展名
        parsed = urlparse(url.split('?')[0])
        base_name = os.path.basename(parsed.path) or None
        return base_name, None

def download_image(url: str, out_path: str) -> bool:
    try:
        # 请求图片
        resp = session.get(url, stream=True, timeout=15)
        resp.raise_for_status()
        # 简单检查 Content-Type 是否为图片
        ct = resp.headers.get("Content-Type", "")
        if not ct.startswith("image"):
            # 不是图片，还是可以尝试保存，但通常我们跳过
            print(f"Warning: URL 可能不是图片 (Content-Type={ct}): {url}")
            return False

        # 写入文件
        with open(out_path, "wb") as f:
            for chunk in resp.iter_content(chunk_size=8192):
                if chunk:
                    f.write(chunk)
        return True
    except Exception as e:
        print(f"下载失败: {url} -> {e}")
        return False
    

def download_image_with_content_type_check(url: str, out_path: str) -> bool:
    try:
        resp = session.get(url, stream=True, timeout=15)
        resp.raise_for_status()
        
        ct = resp.headers.get("Content-Type", "").lower()
        if not ct.startswith("image"):
            print(f"Warning: Non-image content type '{ct}' for {url}")
            return False

        # 根据真实 Content-Type 修正文件扩展名（如果当前扩展名不匹配）
        actual_ext = ".jpg"  # 默认
        if "png" in ct:
            actual_ext = ".png"
        elif "gif" in ct:
            actual_ext = ".gif"
        elif "jpeg" in ct or "jpg" in ct:
            actual_ext = ".jpg"

        # 如果原 out_path 扩展名和实际不符，重命名
        current_ext = os.path.splitext(out_path)[1].lower()
        if current_ext not in [".jpg", ".jpeg"] and actual_ext == ".jpg":
            # 统一用 .jpg
            new_path = os.path.splitext(out_path)[0] + ".jpg"
            out_path = new_path
        elif current_ext != actual_ext:
            new_path = os.path.splitext(out_path)[0] + actual_ext
            out_path = new_path

        # 写入文件
        with open(out_path, "wb") as f:
            for chunk in resp.iter_content(chunk_size=8192):
                if chunk:
                    f.write(chunk)
        return True

    except Exception as e:
        print(f"下载失败: {url} -> {e}")
        return False

# 遍历每一行，逐个下载
count_ok = 0
count_skip = 0
count_fail = 0

for idx, raw in enumerate(df[COLUMN_NAME].fillna("").astype(str)):
    url = raw.strip().strip('\'"')
    if not url:
        count_skip += 1
        continue

    # 有些列可能包含多张图或空格分隔，假设是单个 url（如你说明），如果包含多个可扩展处理
    # 处理常见的编码/空白
    url = url.strip()
    # 获取候选基础名与扩展
    base_name_candidate, ext = get_name_and_ext_from_url(url)

    # 如果没有扩展名，尝试用 HEAD 或 GET 获得 Content-Type 推断扩展
    if ext is None:
        ext = ".jpg"
        try:
            head = session.head(url, timeout=10, allow_redirects=True)
            ctype = head.headers.get("Content-Type", "")
            if ctype:
                if "jpeg" in ctype:
                    ext = ".jpg"
                elif "png" in ctype:
                    ext = ".png"
                elif "gif" in ctype:
                    ext = ".gif"
                else:
                    ext = ".jpg"  # 默认
            else:
                ext = ".jpg"
        except Exception:
            ext = ".jpg"

    # 如果 base_name_candidate 为空，使用行号生成
    if not base_name_candidate:
        base_name_candidate = f"image_{idx+1}"

    # 仅取扩展名前的最后一段做名字（并进行清理）
    base_name = sanitize_filename(base_name_candidate)

    out_file = unique_path(OUT_DIR, base_name, ext)
    success = download_image(url, out_file)
    # success = download_image_with_content_type_check(url, out_file)
    if success:
        count_ok += 1
    else:
        count_fail += 1

    # 可选：打印进度
    if (idx + 1) % 100 == 0:
        print(f"Processed {idx+1} rows: ok={count_ok}, fail={count_fail}, skip={count_skip}")

print("完成。")
print(f"下载成功: {count_ok}, 跳过(空): {count_skip}, 失败: {count_fail}")
```

## 将多个图片文件夹合并（名字去重）

```python
import os
import shutil

# 定义源文件夹和目标文件夹
source_folders = ['mideast&asia_imgs','war_imgs_total']
target_folder = 'war_imgs_total'

# 支持的图片扩展名（小写）
image_extensions = {'.jpg', '.jpeg', '.png', '.gif', '.bmp', '.tiff', '.webp'}

# 确保目标文件夹存在
os.makedirs(target_folder, exist_ok=True)

# 获取 total 中已有的文件名集合（用于快速查重）
existing_files = set(os.listdir(target_folder))

# 遍历每个源文件夹
for folder in source_folders:
    if not os.path.exists(folder):
        print(f"警告：文件夹 '{folder}' 不存在，跳过。")
        continue
        
    for filename in os.listdir(folder):
        src_path = os.path.join(folder, filename)
        
        # 跳过子目录，只处理文件
        if not os.path.isfile(src_path):
            continue
            
        # 检查是否为图片
        _, ext = os.path.splitext(filename)
        if ext.lower() not in image_extensions:
            continue
            
        # 检查 total 中是否已有同名文件
        if filename in existing_files:
            print(f"跳过（已存在）: {filename}")
            continue
            
        # 复制文件到 total
        dst_path = os.path.join(target_folder, filename)
        shutil.copy2(src_path, dst_path)  # copy2 保留元数据
        existing_files.add(filename)  # 更新已存在集合，防止后续重复
        print(f"已复制: {filename}")

print("所有图片合并完成！")
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.image import imread
from pathlib import Path

# 读取 CSV
CSV_PATH = r"D:/GDELP/ur_conflic_1.csv"
df = pd.read_csv(CSV_PATH, dtype=str, low_memory=False, encoding='utf-8')

# 筛选出既有 title 又有有效 image_path 的行
df_valid = df[(df['title'].notna()) & 
              (df['image_path'].notna()) & 
              (df['image_path'].astype(str).str.strip() != '')].copy()

print(f"有效行数（含title和图片路径）: {len(df_valid)}")

# 随机选取最多50行（如果少于50行则全选）
sample_size = min(50, len(df_valid))
df_sample = df_valid.sample(n=sample_size, random_state=42)

print(f"已随机选取 {len(df_sample)} 行")

# 创建图表展示（5行10列的网格）
cols = 5
rows = (len(df_sample) + cols - 1) // cols
fig, axes = plt.subplots(rows, cols, figsize=(20, 4 * rows))
axes = axes.flatten()  # 平展成一维数组方便遍历

for idx, (_, row) in enumerate(df_sample.iterrows()):
    ax = axes[idx]
    title = row['title'][:50] + '...' if len(str(row['title'])) > 50 else row['title']  # 截断长标题
    img_path = row['downloaded_image_path']
    
    try:
        # 检查文件是否存在
        if os.path.exists(img_path):
            img = imread(img_path)
            ax.imshow(img)
            ax.set_title(title, fontsize=10, wrap=True)
        else:
            ax.text(0.5, 0.5, f'文件不存在\n{img_path}', 
                   ha='center', va='center', fontsize=9, wrap=True)
            ax.set_title(title, fontsize=10)
    except Exception as e:
        ax.text(0.5, 0.5, f'加载失败\n{str(e)[:30]}', 
               ha='center', va='center', fontsize=8)
        ax.set_title(title, fontsize=10)
    
    ax.axis('off')

# 隐藏多余的空白坐标轴
for idx in range(len(df_sample), len(axes)):
    axes[idx].axis('off')

plt.tight_layout()
plt.savefig('sample_articles_display.png', dpi=100, bbox_inches='tight')
print("图表已保存为 'sample_articles_display.png'")
plt.show()

# 同时输出采样数据的摘要表格
print("\n采样数据摘要：")
print(df_sample[['title', 'downloaded_image_path']].to_string())
```

```python
import os
from PIL import Image
from pathlib import Path

image_dir = "/opt/data/private/war_news_bias_detection/war_imgs_total"
bad_files = []

for img_path in Path(image_dir).rglob("*.jpg"):
    try:
        with Image.open(img_path) as img:
            img.verify()  # 快速验证格式
    except Exception as e:
        print(f"Bad image: {img_path} - {e}")
        bad_files.append(str(img_path))

print(f"\nFound {len(bad_files)} corrupted images.")
# 可选：删除坏图
# for f in bad_files:
#     os.remove(f)
```

```python
OUTPUT_CSV = "label-gemini-flash-lite-2.5.csv"
print("🧹 正在清理 CSV 中的坏样本...")
initial_count = len(df)
df_clean = df[~df["image_path"].isin(bad_files)]
final_count = len(df_clean)

print(f"📊 原始样本数: {initial_count}")
print(f"📊 清理后样本数: {final_count}")
print(f"🗑️  删除了 {initial_count - final_count} 条坏数据")

# === 步骤 4：保存清理后的 CSV（不包含临时列）===
df_clean.to_csv(OUTPUT_CSV, index=False)

print(f"\n✅ 清理完成！新 CSV 已保存至:\n{OUTPUT_CSV}")
```
